# ViewViewView · JupyterLab 实时演示

这是一个**真实运行**的演示：下方单元格会连续执行 800 次模拟故障注入，
观察上窗分类计数（白色）与下窗最新参数实时变化。

- 表头分隔线可**拖动调整列宽**
- CSV 自动落盘，默认文件名 = 本控件创建时刻
- 字段在实例化时通过 `columns=` 自定义

```python
import random
import time

import viewviewview as faultviz

vt = faultviz.ViewWidget(
    live_rows=500,
    columns={
        "state": str,
        "delay": int,
        "pulse": int,
        "reply": str,
        "errcode": int,
    },
)
vt.show()
```

```python
rng = random.Random(20261019)

for index in range(800):
    roll = rng.random()
    if roll < 0.02:
        state, reply, errcode = "TIMEOUT", "open", 7
    elif roll < 0.11:
        state, reply, errcode = "FAIL", rng.choice(["no-reply", "short"]), rng.randint(1, 8)
    else:
        state, reply, errcode = "PASS", "ok", 0
    vt.update(
        state=state,
        delay=rng.randint(100, 2000),
        pulse=rng.randint(10, 50),
        reply=reply,
        errcode=errcode,
    )
    time.sleep(0.01)

print("rows recorded:", vt.rows_written)
print("csv path      :", vt.csv_path)
```
